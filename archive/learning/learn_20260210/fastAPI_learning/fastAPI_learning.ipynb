# 1. 先理解一个网页应用的基本结构

想象你要开一家餐厅：

```
客人点餐（请求）
    ↓
```

# 4. 什么是"路由"？

````python
print("=" * 60)
print("路由 (Router)")
print("=" * 60)
print('''
路由就是 "哪个地址对应哪个处理函数"。

类比：
┌───────────────────────────────────────┐
│ 餐厅菜单                                │
├───────────────────────────────────────┤
│ 红烧肉    28元   →  厨房做红烧肉       │
│ 宫保鸡丁  22元   →  厨房做宫保鸡丁     │
│ 米饭      3元    →  厨房盛米饭         │
└───────────────────────────────────────┘

FastAPI 路由：
┌───────────────────────────────────────┐
│ @app.get("/hello")    →  say_hello函数 │
│ @app.post("/items")  →  create_item   │
│ @app.get("/items/{id}") → get_item    │
└───────────────────────────────────────┘
''')
````

````python
print("路由的组成部分")
print("=" * 60)
print('''
┌───────────────────────────────────────┐
│  @app.get("/items/{item_id}")          │
│  ───────  ───────────
      ──────────────    │
│    │              │                    │
│  装饰器       路径地址                │
│ (决定方法)   (决定处理哪个请求)       │
└───────────────────────────────────────┘
''')
````

### 装饰器是什么？

````python
# 装饰器就是 @xxx 这种语法糖
# 它本质上是一个函数，把另一个函数包装一下

def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("开始执行...")
        result = func(*args, **kwargs)
        print("执行结束")
        return result
    return wrapper

@my_decorator
def say_hello():
    print("你好！")

say_hello()

print('''
输出：
开始执行...
你好！
执行结束

@app.get() 装饰器做了类似的事情：
- 告诉 FastAPI "这个函数对应这个路径"
- FastAPI 会在请求到来时调用这个函数
''')
````

### 路径参数的类型校验

````python
# Path 参数校验
from fastapi import Path

print('''
@app.get("/items/{item_id}")
async def get_item(
    item_id: int = Path(..., ge=1)  # ge=1 表示 >= 1
):
    return {"item_id": item_id}

# 访问 /items/abc → 422 错误（不是数字）
# 访问 /items/0   → 422 错误（小于1）
# 访问 /items/1   → 正常

# 常用校验：
# ge = greater than or equal (>=)
# gt = greater than (>)
# le = less than or equal (<=)
# lt = less than (<)
''')
````

# 9. lifespan 生命周期

````python
from contextlib import asynccontextmanager

print("=" * 60)
print("lifespan 生命周期")
print("=" * 60)

print('''
# lifespan 就像餐厅的开店打烊流程

@asynccontextmanager
async def lifespan(app: FastAPI):
    # ========== 餐厅开门 ==========
    print("餐厅开门了")
    await init_database()    # 初始化数据库
    await init_cache()       # 初始化缓存
    
    yield  # 营业时间，请求在这里处理
    
    # ========== 餐厅打烊 ==========
    print("餐厅打烊了")
    await save_state()       # 保存状态
    await cleanup()          # 清理资源

app = FastAPI(lifespan=lifespan)
''')

print('''
执行流程：
启动时 → init_database → init_cache → yield → 请求处理 → save_state → cleanup → 关闭
''')
````

# 10. 中间件（请求拦截）

````python
from fastapi import FastAPI, Request
import time

print("=" * 60)
print("中间件")
print("=" * 60)

print('''
# 中间件 = 请求拦截器

app = FastAPI()

@app.middleware("http")
async def add_process_time(request: Request, call_next):
    start_time = time.time()
    response = await call_next(request)
    process_time = time.time() - start_time
    response.headers["X-Process-Time"] = str(process_time)
    return response
''')
````

````python
print("中间件执行顺序（洋葱模型）")
print("=" * 60)
print('''
┌───────────────────────────────────────┐
│   请求                                │
│     ↓                                │
│   中间件1 进入                        │
│     ↓                                │
│   中间件2 进入                        │
│     ↓                                │
│   ════════════════════                │
│        路由处理函数                   │
│   ════════════════════                │
│     ↓                                │
│   中间件2 退出                        │
│     ↓                                │
│   中间件1 退出                        │
│     ↓                                │
│   响应                                │
└───────────────────────────────────────┘
''')
````

## 练习2：待办事项 API 参考答案

创建文件 `todos.py`：

````python
"""
待办事项 API - 参考答案
保存为 todos.py
运行: uvicorn todos:app --reload
"""

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from typing import Optional

app = FastAPI(title="待办事项 API")

# 数据模型
class TodoCreate(BaseModel):
    title: str
    description: Optional[str] = None

class TodoUpdate(BaseModel):
    completed: bool

class Todo(BaseModel):
    id: int
    title: str
    description: Optional[str] = None
    completed: bool = False

# 模拟数据库
todos_db = {}
next_id = 1

# POST /todos - 创建待办
@app.post("/todos", response_model=Todo, status_code=201)
async def create_todo(todo: TodoCreate):
    global next_id
    new_todo = Todo(
        id=next_id,
        title=todo.title,
        description=todo.description
    )
    todos_db[next_id] = new_todo
    next_id += 1
    return new_todo

# GET /todos - 获取所有待办
@app.get("/todos", response_model=list[Todo])
async def get_todos():
    return list(todos_db.values())

# GET /todos/{todo_id} - 获取单个待办
@app.get("/todos/{todo_id}", response_model=Todo)
async def get_todo(todo_id: int):
    if todo_id not in todos_db:
        raise HTTPException(status_code=404, detail="待办不存在")
    return todos_db[todo_id]

# PUT /todos/{todo_id} - 更新待办
@app.put("/todos/{todo_id}", response_model=Todo)
async def update_todo(todo_id: int, update: TodoUpdate):
    if todo_id not in todos_db:
        raise HTTPException(status_code=404, detail="待办不存在")
    
    todo = todos_db[todo_id]
    todo.completed = update.completed
    todos_db[todo_id] = todo
    return todo

# DELETE /todos/{todo_id} - 删除待办
@app.delete("/todos/{todo_id}", status_code=204)
async def delete_todo(todo_id: int):
    if todo_id not in todos_db:
        raise HTTPException(status_code=404, detail="待办不存在")
    del todos_db[todo_id]

# 测试:
# POST /todos  Body: {"title": "买水果", "description": "苹果和香蕉"}
# GET /todos
# GET /todos/1
# PUT /todos/1  Body: {"completed": true}
# DELETE /todos/1
````

## 练习3：用户登录 API 参考答案

创建文件 `auth.py`：

````python
"""
用户登录 API - 参考答案
保存为 auth.py
运行: uvicorn auth:app --reload
"""

from fastapi import FastAPI, HTTPException, Depends, Header
from pydantic import BaseModel, EmailStr
from typing import Optional
import hashlib
import secrets

app = FastAPI(title="用户登录 API")

# 数据模型
class UserRegister(BaseModel):
    name: str
    email: EmailStr
    password: str

class UserLogin(BaseModel):
    email: EmailStr
    password: str

class User(BaseModel):
    id: int
    name: str
    email: str

class Token(BaseModel):
    access_token: str
    token_type: str = "bearer"

# 模拟数据库
users_db = {}  # email -> User
tokens_db = {}  # token -> user_id
next_user_id = 1

# 密码哈希函数
def hash_password(password: str) -> str:
    return hashlib.sha256(password.encode()).hexdigest()

# 生成随机 token
def generate_token() -> str:
    return secrets.token_urlsafe(32)

# POST /auth/register - 注册用户
@app.post("/auth/register", response_model=User, status_code=201)
async def register(user: UserRegister):
    global next_user_id
    
    # 检查邮箱是否已存在
    if user.email in users_db:
        raise HTTPException(status_code=400, detail="邮箱已被注册")
    
    # 创建用户
    new_user = User(
        id=next_user_id,
        name=user.name,
        email=user.email
    )
    users_db[user.email] = {
        "user": new_user,
        "password_hash": hash_password(user.password)
    }
    next_user_id += 1
    return new_user

# POST /auth/login - 登录
@app.post("/auth/login", response_model=Token)
async def login(user: UserLogin):
    # 检查用户是否存在
    if user.email not in users_db:
        raise HTTPException(status_code=401, detail="邮箱或密码错误")
    
    # 检查密码
    stored = users_db[user.email]
    if stored["password_hash"] != hash_password(user.password):
        raise HTTPException(status_code=401, detail="邮箱或密码错误")
    
    # 生成 token
    token = generate_token()
    tokens_db[token] = stored["user"].id
    
    return Token(access_token=token)

# 获取当前用户（依赖注入）
def get_current_user(token: str = Header(...)) -> User:
    if token not in tokens_db:
        raise HTTPException(status_code=401, detail="无效的token")
    user_id = tokens_db[token]
    for u in users_db.values():
        if u["user"].id == user_id:
            return u["user"]
    raise HTTPException(status_code=401, detail="用户不存在")

# GET /profile - 获取当前用户信息
@app.get("/profile", response_model=User)
async def get_profile(current_user: User = Depends(get_current_user)):
    return current_user

# 测试:
# 1. POST /auth/register  Body: {"name": "张三", "email": "zhangsan@example.com", "password": "123456"}
# 2. POST /auth/login     Body: {"email": "zhangsan@example.com", "password": "123456"}
# 3. GET /profile        Header: Authorization: Bearer <返回的token>
````
